# NOW Index — Quickstart Guide

This notebook demonstrates how to use the NOW Index API to fetch rankings, scores, and asset data.

```python
import requests
import json

BASE_URL = "http://localhost:8000"  # Change to your API URL
```

## 1. Health Check

```python
r = requests.get(f"{BASE_URL}/api/health")
print(json.dumps(r.json(), indent=2))
```

## 2. Get Top 10 Rankings

```python
r = requests.get(f"{BASE_URL}/api/top10")
data = r.json()
for i, result in enumerate(data['results'], 1):
    print(f"{i:2d}. {result['ticker']:6s} {result['score']:6.1f}  {result['name']}")
```

## 3. Company Profile

```python
r = requests.get(f"{BASE_URL}/api/company/AAPL")
data = r.json()
print(f"Company: {data['now_score']['name']}")
print(f"Ticker:  {data['now_score']['ticker']}")
print(f"NOW Score: {data['now_score']['score']:.1f}")
print(f"Rank:     #{data['now_score']['rank']}")
print(f"\nFactor Breakdown:")
for k, v in data['now_score']['factors'].items():
    if k != 'total':
        print(f"  {k:20s}: {v:.1f}")
```

## 4. Compare Assets

```python
r = requests.get(f"{BASE_URL}/api/compare?tickers=AAPL,MSFT,GOOGL")
data = r.json()
for result in data['results']:
    s = result['now_score']
    print(f"{s['ticker']:6s} | Score: {s['score']:6.1f} | Rank: #{s['rank']:3d} | Quality: {s['factors']['quality']:5.1f} | Value: {s['factors']['value']:5.1f}")
```

## 5. Search Assets

```python
r = requests.get(f"{BASE_URL}/api/search?q=APPLE")
data = r.json()
print(f"Found {data['count']} results:")
for result in data['results']:
    print(f"  {result['asset']['ticker']:10s} {result['asset']['name']}")
```

## 6. Get Leaderboard Category

```python
r = requests.get(f"{BASE_URL}/api/leaderboard/highest_momentum")
data = r.json()
print(f"Category: {data['category']}")
for i, s in enumerate(data['results'][:5], 1):
    print(f"{i}. {s['ticker']:6s} Score: {s['score']:.1f} Momentum: {s['factors']['momentum']:.1f}")
```

## 7. Get Platform Stats

```python
r = requests.get(f"{BASE_URL}/api/stats")
print(json.dumps(r.json(), indent=2))
```
